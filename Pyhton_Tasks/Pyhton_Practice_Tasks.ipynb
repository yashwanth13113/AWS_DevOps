```python

def fibonacci_series(n):
    series = []
    a, b = 0, 1
    for _ in range(n):
        series.append(a)
        a, b = b, a + b
    return series
print(fibonacci_series(10))


def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
def fibonacci_series(n):
    return [fibonacci(i) for i in range(n)]
print(fibonacci_series(10))



    

```

```python
#Project #1 – Python Magic 8 Ball
import sys
import random

# initial variables
questions = True

# list of responses
responses = [
    "It is certain",
    "You may rely on it",
    "As I see it, yes",
    "Outlook good",
    "It is decidedly so",
    "Without a doubt",
    "Yes, definitely"
]

while questions:
    # user question input
    ques = input("Ask a question (press ENTER to quit): ")

    # exit condition
    if ques == "":
        sys.exit()

    # random response generator
    magic_response = responses[random.randint(0, 6)]

    # print response
    print(magic_response)
```
